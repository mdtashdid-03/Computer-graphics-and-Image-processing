```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Upload/read image
image = cv2.imread('/images.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
# Image dimensions
h, w = image.shape[:2]
# Translation values
tx = 100
ty = 50
# Translation matrix
M = np.float32([
[1, 0, tx],
[0, 1, ty]
])
# Apply transformation
translated = cv2.warpAffine(image, M, (w, h))# Display
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('//images.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
h, w = image.shape[:2]
# Scaling factors
sx = 4
sy = 3
# Scaling matrix
M = np.float32([
[sx, 0, 0],
[0, sy, 0]
])
scaled = cv2.warpAffine(
image,
M,
(int(w * sx), int(h * sy)))
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
image = cv2.imread('/images.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
h, w = image.shape[:2]
# Center of image
center = (w // 2, h // 2)
# Rotation angle
angle = 90
# Rotation matrix
M = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated = cv2.warpAffine(
image,
M,
(w, h)
)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(rotated)
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/images.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Reflection about X-axis
reflectedx = cv2.flip(image, 0)
reflectedy = cv2.flip(image, 1)

plt.figure(figsize=(25, 20))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(reflectedx)
plt.title("Reflection about X-axis")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(reflectedy)
plt.title("Reflection about Y-axis")
plt.axis("off")
plt.show()
```
